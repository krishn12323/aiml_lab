```python
# activity 1: create array
import numpy as np

a=np.array([10,20,30,40])
print(a)

b=np.arange(1,11)
print(b)

c=np.linspace(0,1,5)
print(c)
```

```python
# A2: matrix operation

A=np.array([[1,2],
           [3,4]])
B=np.array([[5,4],
           [7,8]])

print(A+B)
print(A*B)
print(A-B)
print(A@B)

```

```python
#A3:Statistics
marks= np.array([70,65,80,90,88])

print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.std(marks))
```
